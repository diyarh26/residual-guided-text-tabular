```python
import re
from pathlib import Path
from google.colab import drive

HELPER_SHA = "b6a34226d88664c50db42647e312598e5bf59017"
if re.fullmatch(r"[0-9a-f]{40}", HELPER_SHA) is None:
    raise ValueError("Use the verified full transfer-helper commit.")
drive.mount("/content/drive")
ORIGINAL_RUN = Path("/content/drive/MyDrive/ReFuTab/runs/phase4-checkpointed-20261020T145755Z-dd78ef55")
if not (ORIGINAL_RUN / "manifest.json").is_file():
    raise FileNotFoundError(f"Original completed run missing: {ORIGINAL_RUN}. Mount its original Drive or correct ORIGINAL_RUN.")
```

```python
import subprocess
import sys
import urllib.request
from uuid import uuid4
from google.colab import files

helper = Path("/content/refutab_audit_transfer.py")
url = f"https://raw.githubusercontent.com/diyarh26/residual-guided-text-tabular/{HELPER_SHA}/src/refutab/audit_transfer.py"
urllib.request.urlretrieve(url, helper)
archive = Path("/content") / ("refutab-pilot-audit-" + uuid4().hex[:8] + ".zip")
subprocess.run([sys.executable, str(helper), "--run-directory", str(ORIGINAL_RUN), "--output", str(archive)], check=True)
print("Private audit inputs only. Original run is unchanged:", archive)
files.download(str(archive))
```
